```python
import numpy as np
import matplotlib.pyplot as plt

def main():
    X_mu = np.array([6, 4, 4, 5, 3, 4, 5, 0, 0, 0, 0]).reshape(-1, 1)
    N = len(X_mu)

    k = np.arange(N).reshape(-1, 1)
    mu = np.arange(N).reshape(1, -1)
    K = k @ mu

    W = np.exp(1j * (2 * np.pi / N) * K)

    x_k = (1 / N) * (W @ X_mu)

    print(f"Number of samples N = {N}\n")
    print("Matrix K:")
    print(K)
    print("\nMatrix W (rounded):")
    print(np.round(W, 2))

    plt.figure(figsize=(10, 6))
    plt.stem(np.arange(N), x_k.real.flatten(), linefmt='b-', markerfmt='bo', label='Real part')
    plt.stem(np.arange(N), x_k.imag.flatten(), linefmt='r--', markerfmt='rx', label='Imaginary part')
    plt.title(f'Synthesized signal x[k] (N={N}) - Variant 5')
    plt.xlabel('Sample index k')
    plt.ylabel('Amplitude')
    plt.legend()
    plt.grid(True)
    plt.show()

if __name__ == '__main__':
    main()
```
